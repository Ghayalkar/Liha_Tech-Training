```python
def frequency(a):
    b = {}
    
    for x in a:
        if x in b:
            b[x] = b[x] + 1
        else:
            b[x] = 1
    
    return b

a = ['A', 'B', 'A', 'C', 'B', 'A']
print(frequency(a))
```

```python
def digit_sum(a):
    if a == 0:
        return 0
    return a % 10 + digit_sum(a // 10)

a = digit_sum(5832)
print("Sum of digits =", a)
```

```python
def fibonacci(a):
    if a == 0:
        return 0
    if a == 1:
        return 1
    return fibonacci(a - 1) + fibonacci(a - 2)

a = fibonacci(8)
print("Fibonacci number =", a)
```

```python
def tax(a):
    if a <= 300000:
        return 0
    elif a <= 600000:
        return (a - 300000) * 5 / 100
    else:
        return 300000 * 5 / 100 + (a - 600000) * 10 / 100

a = tax(800000)
print("Tax =", a)
```

```python
def salary(a, b, c):
    def calculate(x, y, z):
        return x + x * y / 100 + x * z / 100
    
    return calculate(a, b, c)

a = salary(30000, 20, 10)
print("Gross salary =", a)
```

```python
def validate(a):
    if len(a) < 8:
        return False, "Password must have at least 8 characters"
    
    if not any(x.isupper() for x in a):
        return False, "Password must contain one uppercase letter"
    
    if not any(x.isdigit() for x in a):
        return False, "Password must contain one digit"
    
    return True, "Valid password"

a, b = validate("Python1234")
print("Status =", a)
print("Reason =", b)
```

```python
def flatten(a):
    b = []

    for x in a:
        if isinstance(x, list):
            b.extend(flatten(x))
        else:
            b.append(x)

    return b

a = [1, [2, [3, 4], 5], [6, 7]]
print(flatten(a))
```

```python
def search(a, target):
    b = 0
    c = len(a) - 1

    while b <= c:
        d = (b + c) // 2

        if a[d] == target:
            return d
        elif a[d] < target:
            b = d + 1
        else:
            c = d - 1

    return -1

a = [2, 5, 8, 12, 16, 23, 38]
print("Index =", search(a, 16))
```

```python
def merge(a):
    a.sort()
    b = [a[0]]

    for x, y in a[1:]:
        if x <= b[-1][1]:
            b[-1] = (b[-1][0], max(b[-1][1], y))
        else:
            b.append((x, y))

    return b

a = [(1, 3), (2, 6), (8, 10), (9, 12)]
print(merge(a))
```

```python
def prefix(a):
    b = a[0]

    for x in a[1:]:
        while not x.startswith(b):
            b = b[:-1]

    return b

a = ['flower', 'flow', 'flight']
print("Longest common prefix =", prefix(a))
```

```python
def anagrams(a):
    b = {}

    for x in a:
        c = ''.join(sorted(x))

        if c not in b:
            b[c] = []

        b[c].append(x)

    return list(b.values())

a = ['eat', 'tea', 'tan', 'ate', 'nat', 'bat']
print(anagrams(a))
```

```python
def frequent(a, k):
    b = {}

    for x in a:
        b[x] = b.get(x, 0) + 1

    c = sorted(b, key=b.get, reverse=True)

    return c[:k]

a = [1, 1, 1, 2, 2, 3]
print(frequent(a, 2))
```

```python
def pascal(n):
    a = [1]

    for x in range(n - 1):
        b = [1]

        for y in range(len(a) - 1):
            b.append(a[y] + a[y + 1])

        b.append(1)
        a = b

    return a

a = pascal(5)
print(a)
```

```python
def calculator(a, b, operation):
    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b
    elif operation == "multiply":
        return a * b
    elif operation == "divide":
        return a / b

a = calculator(20, 5, "divide")
print("Result =", a)
```

```python
def multiply(A, B):
    C = [[0, 0], [0, 0]]

    for i in range(2):
        for j in range(2):
            for k in range(2):
                C[i][j] = C[i][j] + A[i][k] * B[k][j]

    return C

A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]

print(multiply(A, B))
```

```python
def balanced(a):
    b = []
    c = {')': '(', ']': '[', '}': '{'}

    for x in a:
        if x in "([{":
            b.append(x)
        elif x in ")]}":
            if not b or b[-1] != c[x]:
                return False
            b.pop()

    return len(b) == 0

a = balanced("{[()]}")
print(a)
```
