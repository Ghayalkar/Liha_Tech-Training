### Q29: Import matplotlib.pyplot and plot a histogram of `Postal Code`.

### Q30: Plot a bar chart showing the top 10 most frequent values in `VIN (1-10)`.

### Q31: Plot a scatter plot between `Postal Code` and `Model Year`.

### Q32: Group by `County` and plot the sum of `Postal Code` as a pie chart (top 5 categories).


```python
# Write your code below
import matplotlib.pyplot as plt
df.groupby('County')['Postal Code'].sum().nlargest(5).plot(kind='pie')
plt.title('Top 5 County by Sum of Postal Code')
plt.ylabel('')
plt.show()
```

### Q35: Use `.apply()` to create a new column indicating whether `Postal Code` is 'High' (above mean) or 'Low' (below mean).


```python
# Write your code below
m = df['Postal Code'].mean()
df['Postal Code Level'] = df['Postal Code'].apply(lambda x:'High' if x>m else 'Low')
```
