### Q29: Import matplotlib.pyplot and plot a histogram of `Rank`.


```python
# Write your code below
import matplotlib.pyplot as plt

plt.hist(df['Rank'])
plt.xlabel('Rank')
plt.ylabel('Frequency')
plt.title('Histogram of Rank')
plt.show()
```

### Q30: Plot a bar chart showing the top 10 most frequent values in `Name`.


```python
# Write your code below
import seaborn as sns 
a = df['Name'].value_counts().nlargest(10).sort_values(ascending=False)
sns.barplot(x=a.index,y=a.values)
```

### Q31: Plot a scatter plot between `Rank` and `Year`.


```python
# Write your code below
import matplotlib.pyplot as plt
plt.scatter(df['Rank'],df['Year'])
plt.xlabel('Rank')
plt.ylabel('Year')
plt.show()
```

### Q32: Group by `Platform` and plot the sum of `Rank` as a pie chart (top 5 categories).


```python
# Write your code below
import matplotlib.pyplot as plt
df.groupby('Platform')['Rank'].sum().sort_values(ascending=False).head().plot(kind='pie')
plt.title('Top 5 Platform by Sum of Rank')
plt.show()
```

### Q35: Use `.apply()` to create a new column indicating whether `Rank` is 'High' (above mean) or 'Low' (below mean).


```python
# Write your code below
m=df['Rank'].mean()
df['Rank_level'] = df['Rank'].apply(lambda x : 'High' if x>m else 'Low')
```
