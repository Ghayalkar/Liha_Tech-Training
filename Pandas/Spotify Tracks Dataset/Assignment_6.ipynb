### Q29: Import matplotlib.pyplot and plot a histogram of `Unnamed: 0`.

### Q30: Plot a bar chart showing the top 10 most frequent values in `track_id`.


```python
# Write your code below
import seaborn as sns
a = df['track_id'].value_counts().nlargest(10).sort_values(ascending=False)
sns.barplot(x=a.index,y=a.values)
```

### Q31: Plot a scatter plot between `Unnamed: 0` and `popularity`.


```python
# Write your code below
import matplotlib.pyplot as plt
plt.scatter(df['Unnamed: 0'], df['popularity'])
plt.xlabel('unnamed: 0')
plt.ylabel('popularity')
plt.show()
```

### Q32: Group by `artists` and plot the sum of `Unnamed: 0` as a pie chart (top 5 categories).

### Q35: Use `.apply()` to create a new column indicating whether `Unnamed: 0` is 'High' (above mean) or 'Low' (below mean).


```python
# Write your code below
m = df['Unnamed: 0'].mean()
df['Unnamed: 0 level'] = df['Unnamed: 0'].apply(lambda x: 'High' if x>m else "Low")
```
