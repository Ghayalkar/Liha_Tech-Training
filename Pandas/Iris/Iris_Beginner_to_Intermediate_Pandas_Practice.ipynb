## About the Dataset

The **Iris.csv** dataset contains measurements for **150 iris flowers** belonging to three species.

It contains **5 columns**:

| Column | Description |
|---|---|
| `sepal_length` | Sepal length in centimetres |
| `sepal_width` | Sepal width in centimetres |
| `petal_length` | Petal length in centimetres |
| `petal_width` | Petal width in centimetres |
| `species` | Iris species: setosa, versicolor, or virginica |

The dataset has no standard missing values. Because it contains both numerical measurements and a categorical species column, it is useful for practicing data inspection, filtering, grouping, aggregation, and visualization.

**Original dataset:** [UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/53/iris)  
**CSV source:** [seaborn-data](https://github.com/mwaskom/seaborn-data/blob/master/iris.csv)

## 8. Correlation, Visualization, and Export

### Question 37

```python
df.corr(numeric_only=True)
```

### Question 38
Create a **histogram** of the `sepal_length` column with an appropriate title and axis labels.


```python
import matplotlib.pyplot as plt

plt.hist(df['sepal_length'])
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')
plt.show()
```

### Question 39
Create a **bar chart** showing the average `petal_length` for each species.


```python
import matplotlib.pyplot as plt

a = df.groupby('species')['petal_length'].mean()

plt.bar(a.index, a.values)
plt.title('Species and Their Average Petal Length')
plt.xlabel('Species')
plt.ylabel('Average Petal Length')
plt.show()
```

### Question 40
Create a **box plot** comparing `petal_length` across the three species.


```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.boxplot(x='species', y='petal_length', data=df)

plt.title('Petal Length Distribution by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length')
plt.show()
```
