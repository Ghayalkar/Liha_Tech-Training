### Q31: Plot a scatter plot between `(Availability of Boy Toilet) Total Boysâ€™ and Co-educational Schools - All Management ` and `(Availability of Girls Toilet) Total Schools - All Management`.

### Q32: Group by `State` and plot the sum of `(Availability of Boy Toilet) Total Boysâ€™ and Co-educational Schools - All Management ` as a pie chart (top 5 categories).

### Q35: Use `.apply()` to create a new column indicating whether `(Availability of Boy Toilet) Total Boysâ€™ and Co-educational Schools - All Management ` is 'High' (above mean) or 'Low' (below mean).


```python
# Write your code below
col = '(Availability of Boy Toilet) Total Boysâ€™ and Co-educational Schools - All Management '

m = df[col].mean()

df['Availability'] = df[col].apply(lambda x: 'High' if x > m else 'Low')
```
