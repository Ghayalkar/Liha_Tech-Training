### Q29: Import matplotlib.pyplot and plot a histogram of `Severity`.


```python
# Write your code below
import matplotlib.pyplot as plt
plt.hist(df['Severity'])
plt.title('Distribution of Severity')
plt.xlabel('Severity')
plt.ylabel('Frequency')
plt.show()
```

### Q30: Plot a bar chart showing the top 10 most frequent values in `ID`.


```python
# Write your code below
import seaborn as sns
a = df['ID'].value_counts().sort_values(ascending=False).head(10)
sns.barplot(x=a.index,y=a.values)
```

### Q31: Plot a scatter plot between `Severity` and `Start_Lat`.


```python
# Write your code below
import matplotlib.pyplot as plt

plt.scatter(df['Severity'], df['Start_Lat'])
plt.xlabel('Severity')
plt.ylabel('Start_Lat')
plt.title('Severity vs Start_Lat')
plt.show()
```

### Q32: Group by `Source` and plot the sum of `Severity` as a pie chart (top 5 categories).


```python
# Write your code below
import matplotlib.pyplot as plt

df.groupby('Source')['Severity'].sum().nlargest(5).plot(kind='pie')

plt.title('Top 5 Sources by Sum of Severity')
plt.ylabel('')
plt.show()
```

### Q35: Use `.apply()` to create a new column indicating whether `Severity` is 'High' (above mean) or 'Low' (below mean).


```python
# Write your code below
m = df['Severity'].mean()

df['Severity_Level'] = df['Severity'].apply(lambda x: 'High' if x > m else 'Low')
df
```
