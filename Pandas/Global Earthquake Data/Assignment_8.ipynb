### Q35: Use `.apply()` to create a new column indicating whether `Latitude` is 'High' (above mean) or 'Low' (below mean).


```python
# Write your code below
m = df['Latitude'].mean()
df['Latitude level'] = df['Latitude'].apply(lambda x :'High' if x>m else 'Low')
```
