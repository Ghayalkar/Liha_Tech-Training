### Q30: Plot a bar chart showing the top 10 most frequent values in `Province/State`.

### Q31: Plot a scatter plot between `Lat` and `Long`.


```python
# Write your code below
import matplotlib.pyplot as plt

plt.scatter(df['Lat'], df['Long'])
plt.xlabel('Lat')
plt.ylabel('Long')
plt.title('Lat vs Long')
plt.show()
```

### Q32: Group by `Country/Region` and plot the sum of `Lat` as a pie chart (top 5 categories).


```python
# Write your code below
import matplotlib.pyplot as plt
df.groupby('Country/Region')['Lat'].sum().nlargest(5).plot(kind='pie') 
plt.title('Top 5 Country/Region by sum of Lat')
plt.show()
```

### Q35: Use `.apply()` to create a new column indicating whether `Lat` is 'High' (above mean) or 'Low' (below mean).


```python
# Write your code below
m=df['Lat'].mean()
df['Lat Level'] = df['Lat'].apply(lambda x:'High' if x>m else 'Low')
df
```
