### Q29: Import matplotlib.pyplot and plot a histogram of `LandAverageTemperature`.


```python
# Write your code below
import matplotlib.pyplot as plt

plt.hist(df['LandAverageTemperature'], bins=30)
plt.xlabel('Land Average Temperature')
plt.ylabel('Frequency')
plt.title('Histogram of Land Average Temperature')
plt.show()
```

### Q30: Plot a bar chart showing the top 10 most frequent values in `dt`.


```python
# Write your code below
import matplotlib.pyplot as plt

df['dt'].value_counts().head(10).plot(kind='bar')

plt.xlabel('Date')
plt.ylabel('Frequency')
plt.title('Top 10 Most Frequent Dates')
plt.show()
```

### Q31: Plot a scatter plot between `LandAverageTemperature` and `LandAverageTemperatureUncertainty`.


```python
# Write your code below
import matplotlib.pyplot as plt

plt.scatter(df['LandAverageTemperature'], df['LandAverageTemperatureUncertainty'])

plt.xlabel('Land Average Temperature')
plt.ylabel('Land Average Temperature Uncertainty')
plt.title('Temperature vs Temperature Uncertainty')

plt.show()
```

### Q32: Group by `dt` and plot the sum of `LandAverageTemperature` as a pie chart (top 5 categories).


```python
# Write your code below
import matplotlib.pyplot as plt

df.groupby('dt')['LandAverageTemperature'].sum().nlargest(5).plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Top 5 Dates by Sum of Land Average Temperature')
plt.ylabel('')
plt.show()
```

### Q35: Use `.apply()` to create a new column indicating whether `LandAverageTemperature` is 'High' (above mean) or 'Low' (below mean).


```python
# Write your code below
m= df['LandAverageTemperature'].mean()

df['Temperature_Level'] = df['LandAverageTemperature'].apply( lambda x: 'High' if x > m else 'Low')
```
