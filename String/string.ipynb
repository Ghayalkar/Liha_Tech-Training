Write a Python program to count the number of vowels and consonants in a given string.

Write a Python program to count uppercase letters, lowercase letters, digits, and special characters in a
string.

```python
a = (input("enter a string:"))
uppercase_letter = 0
lowercase_letter = 0
digits = 0
special_char = 0

for char in a:
    if char.isalpha() and char.isupper():
         uppercase_letter+=1
    elif char.isalpha() and char.islower():
        lowercase_letter+=1
    elif char.isdigit():
        digits+=1
    else:
        special_char+=1
        
print("upper case letters:",uppercase_letter)  
print("lower case letters:",lowercase_letter)                  
print("digits count:",digits)                  
print("special characters count:",special_char)
```

Write a Python program to count the frequency of each word in a sentence, ignoring letter case.

```python
sentence = input("Enter a sentence: ")

words = sentence.lower().split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

for word, count in frequency.items():
    print(f"{word}: {count}")
```

Write a Python program to compress consecutive repeated characters using character counts.

```python
a = input("Enter string: ")

x = ""
c = 1

for i in range(len(a)):
    if i < len(a) - 1 and a[i] == a[i + 1]:
        c += 1
    else:
        x += a[i] + str(c)
        c = 1

print("Compressed string =", x)
```

Write a Python program to find the longest palindromic substring in a given string.

```python
a = input("Enter string: ")

x = ""

for i in range(len(a)):
    for j in range(i + 1, len(a) + 1):
        y = a[i:j]
        if y == y[::-1] and len(y) > len(x):
            x = y

print("One longest palindromic substring =", x)
```

Write a Python program to encrypt a string using a Caesar cipher with a given shift.

```python
a = input("Enter text: ")
x = int(input("Enter shift: "))

y = ""

for i in a:
    if i.isupper():
        y += chr((ord(i) - 65 + x) % 26 + 65)
    elif i.islower():
        y += chr((ord(i) - 97 + x) % 26 + 97)
    else:
        y += i

print("Encrypted text =", y)
```

Write a Python program to decode a run-length encoded string.

```python
a = input("Enter string: ")

x = ""

for i in range(0, len(a), 2):
    x += a[i] * int(a[i + 1])

print("Decoded string =", x)
```

Write a Python program to group a list of words into anagram groups.

```python
a = eval(input("Enter list: "))

x = {}

for i in a:
    y = "".join(sorted(i))
    if y in x:
        x[y].append(i)
    else:
        x[y] = [i]

print(list(x.values()))
```

Write a Python program to determine whether a string is made by repeating a smaller substring.

```python
a = input("Enter string: ")

for i in range(1, len(a) + 1):
    x = a[:i]
    if x * (len(a) // len(x)) == a:
        print("Repeated pattern =", x)
        print("Repeat count =", len(a) // len(x))
        break
```
