Advance

```python
numbers = {100, 4, 200, 1, 3, 2}

longest = []

for num in numbers:
    if num - 1 not in numbers:
        current = []
        x = num

        while x in numbers:
            current.append(x)
            x += 1

        if len(current) > len(longest):
            longest = current

print("Sequence =", longest)
print("Length =", len(longest))
```

```python
numbers = [2, 7, 11, 15, 1, 8]
target = 9

seen = set()
pairs = set()

for num in numbers:
    complement = target - num

    if complement in seen:
        pairs.add(tuple(sorted((num, complement))))

    seen.add(num)

print("Pairs =", pairs)
```

```python
s = {1, 2, 3}

power_set = [set()]

for element in s:
    new_sets = []

    for subset in power_set:
        new_subset = subset.copy()
        new_subset.add(element)
        new_sets.append(new_subset)

    power_set.extend(new_sets)

print("Number of subsets =", len(power_set))

print("Subsets:")
for subset in power_set:
    print(subset)
```

```python
numbers = [1, 2, 1, 3, 4, 2, 3]
k = 4

result = []

for i in range(len(numbers) - k + 1):
    window = numbers[i:i + k]
    result.append(len(set(window)))

print("Distinct counts =", result)
```

```python
numbers = [1, 2, 2, 4, 5]
n = 5

expected = set(range(1, n + 1))
actual = set(numbers)

duplicate = set()

for i in numbers:
    if numbers.count(i) > 1:
        duplicate.add(i)

missing = expected - actual

print("Duplicate =", duplicate.pop())
print("Missing =", missing.pop())
```

```python
universe = {1, 2, 3, 4, 5}

groups = [
    {1, 2},
    {3, 4},
    {5}
]

combined = set()
valid = True

for group in groups:
    if combined & group:
        valid = False
        break

    combined = combined | group

if combined != universe:
    valid = False

print("Valid disjoint partition =", valid)
```

```python
universe = {1, 2, 3, 4, 5}

S1 = {1, 2, 3}
S2 = {2, 4}
S3 = {3, 4, 5}

sets = {
    "S1": S1,
    "S2": S2,
    "S3": S3
}

names = list(sets.keys())

for i in range(len(names)):
    for j in range(i + 1, len(names)):
        union = sets[names[i]] | sets[names[j]]

        if union == universe:
            print("Smallest covering pair =", (names[i], names[j]))
            break
```
