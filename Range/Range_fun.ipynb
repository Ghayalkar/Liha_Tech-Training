```python
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]

C = [[0, 0], [0, 0]]

for i in range(2):
    for j in range(2):
        for k in range(2):
            C[i][j] = C[i][j] + A[i][k] * B[k][j]

print(C)
```

```python
a = [5, 1, 4, 2, 8]

for i in range(len(a)):
    for j in range(len(a) - i - 1):
        if a[j] > a[j + 1]:
            a[j], a[j + 1] = a[j + 1], a[j]

print(a)
```

```python
n = 30
a = [True] * (n + 1)
a[0] = a[1] = False

for x in range(2, int(n ** 0.5) + 1):
    if a[x]:
        for y in range(x * x, n + 1, x):
            a[y] = False

for x in range(2, n + 1):
    if a[x]:
        print(x, end=" ")
```

```python
n = 5

for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```
