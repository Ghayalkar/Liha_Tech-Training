```python
a = int(input("Enter number: "))

x = a
y = 0
z = len(str(a))

while x > 0:
    y += (x % 10) ** z
    x //= 10

if y == a:
    print("Armstrong number")
else:
    print("Not Armstrong number")
```

```python
a = int(input("Enter n: "))

for i in range(1, a + 1):
    x = i
    y = 0
    z = len(str(i))

    while x > 0:
        y += (x % 10) ** z
        x //= 10

    if y == i:
        print(i, end=" ")
```
