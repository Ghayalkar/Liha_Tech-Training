```python
t = (1, 2, 2, 3, 4, 4, 4)

result = ()

for i in t:
    if t.count(i) > 1 and i not in result:
        result = result + (i,)

print("Repeated values =", result)
```

#Advance

```python
matrix = ((1, 2, 3), (4, 5, 6))

result = tuple(zip(*matrix))

print(result)
```

```python
matrix = ((1, 2, 3), (4, 5, 6))

result = tuple(zip(*matrix[::-1]))

print(result)
```

```python
t1 = (1, 2)
t2 = ('A', 'B')

result = ()

for i in t1:
    for j in t2:
        result = result + ((i, j),)

print(result)
```

```python
t = (1, 2, 3, 4)

result = ()

for i in range(len(t)):
    for j in range(i + 1, len(t)):
        result = result + ((t[i], t[j]),)

print(result)
```

```python
def flatten(t):
    result = ()

    for i in t:
        if isinstance(i, tuple):
            result = result + flatten(i)
        else:
            result = result + (i,)

    return result


t = (1, (2, (3, 4)), 5)

print(flatten(t))
```

```python
students = (
    (101, 'Amit', 85),
    (102, 'Neha', 92),
    (103, 'Ravi', 85)
)

result = tuple(sorted(students, key=lambda x: (x[2], x[1])))

print(result)
```

```python
records = (
    ('Amit', 'DS'),
    ('Neha', 'AI'),
    ('Ravi', 'DS')
)

ds = ()
ai = ()

for name, course in records:
    if course == 'DS':
        ds = ds + (name,)
    elif course == 'AI':
        ai = ai + (name,)

print("DS =", ds)
print("AI =", ai)
```

```python
t = (1, 2, 3, 4, 5)
n = 2

n = n % len(t)

result = t[n:] + t[:n]

print(result)
```

```python
t = (1, 2, 5, 3, 4, 6, 7, 2)

longest = ()
current = (t[0],)

for i in range(1, len(t)):
    if t[i] > t[i - 1]:
        current = current + (t[i],)
    else:
        if len(current) > len(longest):
            longest = current
        current = (t[i],)

if len(current) > len(longest):
    longest = current

print("Subsequence =", longest)
print("Length =", len(longest))
```

```python
t1 = (1, 3, 5, 7)
t2 = (2, 4, 6, 8)

i = 0
j = 0
result = ()

while i < len(t1) and j < len(t2):
    if t1[i] < t2[j]:
        result = result + (t1[i],)
        i += 1
    else:
        result = result + (t2[j],)
        j += 1

while i < len(t1):
    result = result + (t1[i],)
    i += 1

while j < len(t2):
    result = result + (t2[j],)
    j += 1

print(result)
```
