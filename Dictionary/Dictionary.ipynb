```python
a = {'student': {'name': 'Amit', 'marks': {'Python': 90}}}
b = {}

def flatten(x, key=""):
    for k, v in x.items():
        a = key + "." + k if key else k
        if isinstance(v, dict):
            flatten(v, a)
        else:
            b[a] = v

flatten(a)

print(b)
```

```python
D1 = {'a': 1, 'b': {'x': 10}}
D2 = {'b': {'y': 20}, 'c': 3}

def merge(a, b):
    for key, value in b.items():
        if key in a and isinstance(a[key], dict) and isinstance(value, dict):
            merge(a[key], value)
        else:
            a[key] = value

merge(D1, D2)

print(D1)
```

```python
a = {'A': 95, 'B': 82, 'C': 95, 'D': 70}
b = {}

for key, value in a.items():
    b.setdefault(value, []).append(key)

values = sorted(b, reverse=True)

for i, value in enumerate(values, 1):
    print("Rank", i, ":", ", ".join(b[value]), "(", value, ")")
```

```python
capacity = 3
requests = ['A', 'B', 'C', 'A', 'D']

a = {}

for x in requests:
    if x in a:
        del a[x]
    a[x] = True

    if len(a) > capacity:
        del a[next(iter(a))]

print(list(a.keys()))
```
