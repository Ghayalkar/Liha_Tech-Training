```python
days = ['mon','tue','wed','thur','fri']
temp = [25,31,34,27,29]
```

```python
plt.plot(days,temp)
```

```python
plt.plot(days,temp,'r--o')
```

```python
plt.Figure(figsize=(3,3))
plt.plot(days,temp,'k--o',label='temp.trend')
plt.xlabel('days')
plt.ylabel('temperature')
plt.title('temperature analysis')
plt.legend()
plt.savefig('temp.png',dpi=350)
```
